## 6. How to Install Python on Your Own Machine

If you want to run these labs locally on your computer instead of using Google Colab, follow the instructions below for your operating system:

### Windows
1. **Download:** Go to the official [Python Downloads page](https://www.python.org/downloads/) and download the latest stable installer for Windows.
2. **Run Installer:** Double-click the downloaded `.exe` file.
3. **CRITICAL STEP:** Check the box at the bottom that says **"Add python.exe to PATH"** before clicking **Install Now**. If you miss this step, your command prompt won't recognize Python.
4. **Verify:** Open your Command Prompt (`cmd`) and type:
   ```bash
   python --version
   ```

### macOS
* **Option A: Official Installer (Easiest)**
  1. Go to [Python Downloads](https://www.python.org/downloads/) and download the macOS installer.
  2. Open and run the package file, following the on-screen prompts.
* **Option B: Homebrew (Recommended for developers)**
  1. Open Terminal and run:
     ```bash
     brew install python
     ```
* **Verify:** Open your Terminal and type:
   ```bash
   python3 --version
   ```

### Linux (Ubuntu / Debian-based)
1. Open your Terminal and update your package lists:
   ```bash
   sudo apt update
   ```
2. Install Python 3 and pip (the package manager):
   ```bash
   sudo apt install python3 python3-pip -y
   ```
3. **Verify:** Run:
   ```bash
   python3 --version
   ```
