## 1. Python History and How Python Runs

### 1.1 A Brief History of Python

Python was created by **Guido van Rossum** in the late 1980s at CWI, which is the national research institute for mathematics and computer science, in Netherlands. Its first public release appeared in **1991**. The name comes from the British comedy group *Monty Python*, not from the snake.

Python was designed to emphasize:

- Readability
- Simplicity
- Developer productivity
- Portability
- A large standard library

Python 2 was released in 2000, followed by Python 3 in 2008. Python 3 improved several aspects of the language but was not fully backward compatible with Python 2. Official support for Python 2 ended in 2020, so modern projects should use Python 3.

Today, Python is widely used for automation, DevOps, web development, testing, systems administration, data science, and artificial intelligence.

### 1.2 Compiled vs. Interpreted Languages

A **compiled language** commonly uses a compiler to translate source code into native machine code before execution. The generated executable can run directly on the processor. Languages such as C and C++ commonly use ahead-of-time compilation.

Typical advantages include:
- Fast execution
- Early detection of certain errors
- No interpreter required at runtime

However, the program usually must be recompiled for different operating systems or processor architectures.

An **interpreted language** uses an interpreter or runtime environment to execute the program. This allows developers to run and test code without first creating a standalone native executable.

Typical advantages include:

- Faster development and testing
- Interactive execution
- Easier portability between supported systems

However, interpreted programs may execute more slowly and require the appropriate interpreter to be installed.

### 1.3 How Python Executes Code

Python is commonly described as an interpreted language, but its execution model includes both compilation and interpretation. The important difference is what it is compiled into.

A C compiler translates source code directly into machine code that the processor can execute. But in the standard **CPython** implementation:

1. Python reads the `.py` source file.
2. The source code is compiled into an intermediate format called **bytecode**, which contains lower-level, portable instructions for the Python virtual machine.
3. The Python Virtual Machine (PVM), which is the core component of the Python runtime environment and acts as an interpreter engine, executes the bytecode instructions.
4. Python may cache bytecode in `__pycache__` as `.pyc` files. The `.pyc` file allows Python to reuse valid bytecode instead of recompiling the source during a later import.
5. The operating system and processor ultimately perform the actual work.

Therefore, Python does not normally compile directly into native machine code like C. It uses both processes:
- Compilation: `.py` source code becomes Python `bytecode`.
- Interpretation: the Python runtime engine (PVM) executes that `bytecode`.

```text
Python source code (.py)
          ↓
       Bytecode
          ↓
Python Virtual Machine
          ↓
     Program output
```

### 2.1 Versions

A computer can have multiple Python versions installed at the same time. Always verify which version and executable your terminal, script, or development environment is using.

On Linux and macOS:

python -c "import sys; print(sys.executable)"
python -c "import sys; print(sys.version)"

### 2.2 Common Ways to Run Python

Python code can be executed in several ways. The following are the three most common:

#### 1. Interactive REPL

Enter `python` or `python3` in a terminal to start the **Read–Evaluate–Print Loop (REPL)**:

```bash
python3
```

The REPL executes one statement at a time. It is useful for quick experiments, testing expressions, and exploring Python objects.

```text
>>> name = "Amir"
>>> print(f"Hello, {name}")
Hello, Amir
```

Use `exit()` to leave the REPL.

#### 2. Python Script

Save Python code in a `.py` file and pass it to the interpreter:

```bash
python3 myscript.py
```

Python normally executes the file from top to bottom and exits when it finishes. Scripts are commonly used for automation, command-line tools, scheduled tasks, CI/CD pipelines, and production applications.

#### 3. Jupyter Notebook

A Jupyter notebook divides code into independently executable **cells**. A selected Python **kernel** executes the cells and preserves variables and other state between executions.

Notebooks are useful for:

- Interactive demonstrations
- Data exploration and visualization
- Combining documentation with executable examples
- Testing small portions of code

Because cells can be executed in any order, the notebook’s current state may not match the visual order of its cells. When results become confusing, restart the kernel and run all cells from the beginning.

> **Important:** The REPL, script, and notebook do not necessarily use the same Python executable. Check the active interpreter with:

```python
import sys
print(sys.executable)
```

In DevOps, scripts are typically used for automation and production workflows, notebooks for demonstrations and exploratory analysis, and the REPL for quick testing and troubleshooting.

### 2.3 `PATH` and `PYTHONPATH`

`PATH` and `PYTHONPATH` are environment variables, but they serve different purposes.

#### `PATH`: Finding the Python Executable

It tells the operating system’s shell where to search for executable programs. When you enter:

```bash
python
```

the shell searches the directories in `PATH` from left to right and runs the first matching executable it finds. If a Python 3.9 directory appears before a Python 3.11 directory, the `python` command may start Python 3.9.

View `PATH` from bash:

```bash
echo $PATH
```

View `PATH` from Python:

```python
import os

print(os.environ.get("PATH"))
```

Find the selected Python executable:

```bash
which python3      # Linux and macOS
```

You can also check it from Python:

```python
import sys
print(sys.executable)
```

#### `PYTHONPATH`: Finding Python Modules

`PYTHONPATH` tells Python about additional directories to search when importing modules.

For example:

```bash
export PYTHONPATH="/opt/company/modules"
```

Python adds these directories to `sys.path`, which contains the locations searched by statements such as:

```python
import mymodule
```

Inspect the module search path with:

```python
import sys
print(sys.path)
```

The main difference is:

- `PATH` helps the shell locate the **Python executable**.
- `PYTHONPATH` helps Python locate **modules and packages**.

> **Best practice:** Avoid setting a global `PYTHONPATH` unless necessary because it can cause unexpected import conflicts. Prefer virtual environments and properly installed packages for project dependencies.

These variables are especially important on shared servers, development machines, containers, and CI/CD runners, where multiple Python installations or project environments may exist.

````python
import os
!echo "PATH from bash: $PATH"
print("PATH from Python:", os.environ.get("PATH"))
!which python3
````

````python
%%bash

# Create a directory containing a simple Python module
mkdir -p /tmp/company_modules

cat > /tmp/company_modules/mymodule.py <<'PYTHON'
def greet(name):
    return f"Hello, {name}!"
PYTHON

# Append the directory to PYTHONPATH
export PYTHONPATH="${PYTHONPATH:+$PYTHONPATH:}/tmp/company_modules"
# If PYTHONPATH is already set and non-empty, expand to its current value followed by :

echo "PYTHONPATH=$PYTHONPATH"

# Start Python with the updated PYTHONPATH
python3 - <<'PYTHON'
import sys
import mymodule

print("\nModule search paths:")
print(sys.path)

print("\nImported module location:")
print(mymodule.__file__)

print("\nFunction result:")
print(mymodule.greet("DevOps class"))
PYTHON
````

### 2.4 Python Syntax Essentials

#### Indentation Defines Code Blocks

Python uses indentation instead of braces to define blocks of code. The standard convention is **four spaces per indentation level**. Mixing tabs and spaces can cause indentation errors.

```python
temperature = 25

if temperature > 20:
    print("Warm")
    print("No jacket required")

print("Finished")
```

#### Write One Statement per Line

Each statement should normally appear on its own line:

```python
name = "router01"
status = "online"
```

Python permits semicolons, but they reduce readability:

```python
name = "router01"; status = "online"  # Valid, but not recommended
```

#### Continue Long Expressions

Expressions enclosed in parentheses, brackets, or braces can continue naturally across multiple lines:

```python
total = (
    first_value
    + second_value
    + third_value
)
```

```python
servers = [
    "web01",
    "web02",
    "database01",
]
```

A backslash can also continue a line, but implicit continuation inside brackets is preferred:

```python
total = first_value + \
        second_value
```

#### Comments and Docstrings

Comments begin with `#` and explain the purpose or reasoning behind code:

```python
# Retry the connection up to three times
max_retries = 3
```

A **docstring** documents a module, function, class, or method. It is written as the first statement inside the item being documented:

```python
def check_server(hostname):
    """Return whether the specified server is reachable."""
    return True
```

Unlike ordinary comments, docstrings are available at runtime:

```python
print(check_server.__doc__)
```

#### Python Is Case-Sensitive

Names that differ in capitalization are treated as different identifiers:

```python
total = 10
Total = 20
TOTAL = 30
```

Python keywords are also lowercase. For example, use `True`, `False`, and `None` with their exact capitalization.

### 2.5 PEP 8 Naming Conventions

**PEP 8** is Python’s official style guide. Consistent naming makes code easier to read, review, and maintain.

| Kind | Convention | Examples |
|---|---|---|
| Variable | `lower_snake_case` | `port_count`, `server_status` |
| Function | `lower_snake_case` | `build_report()`, `check_server()` |
| Constant | `UPPER_SNAKE_CASE` | `MAX_RETRIES`, `DEFAULT_PORT` |
| Class | `CapWords` or `PascalCase` | `DeviceProfile`, `ServerMonitor` |
| Module | Short `lowercase` name; underscores if helpful | `utils.py`, `network_tools.py` |
| Package | Short `lowercase` name | `automation`, `monitoring` |
| Internal name | Leading underscore | `_temporary_value`, `_helper()` |

Additional conventions include:

- Use descriptive names such as `retry_count` instead of unclear names such as `rc`.
- Avoid names that differ only by capitalization.
- Do not use Python keywords such as `class`, `for`, or `return` as identifiers.
- Avoid single-letter names except for small counters or short mathematical expressions.
- A leading underscore indicates that a name is intended for internal use.


````python
def area(radius):
    """Return the area of a circle with the given radius"""
    return 3.14 * radius ** 2

print(area(5))
print(area.__doc__)   # a docstring is a real, accessible object attribute
````

## 3. Variables and Basic Data Types

### 3.1 Assignment
The most important thing to internalise about Python: a variable is **a name bound to an object**, not a typed box. Assignment does not copy anything — it binds a name to whatever the right-hand side evaluates to.

````python
x = 10                                        # simple
a = b = c = 0                                 # chained: three names, ONE object
name, age, city = "Amir", 40, "Ottawa"        # multiple / tuple unpacking on one line
print(x, a, b, c, name, age, city)
````

### 3.2 Dynamic Typing, `id()`, and `type()`
Names are not typed. The object they point to has a type. You can rebind the same name to a different type at any time.

````python
v = 42
print(v, type(v), id(v))
# id is an integer that uniquely identifies that object during its lifetime.
# In CPython, id is typically the object’s memory address.

v = "forty-two"                # same name, new object, new id
print(v, type(v), id(v))
````

### 3.3 Object References
Because assignment does not copy, two names can end up pointing to the same mutable object. A change through one is visible through the other.

````python
a = [1, 2, 3]
b = a                          # b and a refer to the SAME list object
b.append(4)
print("a =", a)                 # [1, 2, 3, 4]  — the mutation is visible through a too
print("id(a) == id(b)?", id(a) == id(b))
````

### 3.4 Basic Data Types
You will use most of these every week.

| Type | Example | Category |
|------|---------|----------|
| `int` | `42` | number |
| `float` | `3.14` | number |
| `complex` | `1+2j` | number |
| `bool` | `True`, `False` | boolean (subclass of `int`, surprisingly) |
| `str` | `"hello"` | text |
| `NoneType` | `None` | absence of a value |
| `bytes` | `b"hi"` | raw binary (immutable) |
| `bytearray` | `bytearray(b"hi")` | raw binary (mutable) |

````python
for value in [42, 3.14, 1 + 2j, True, "hi", None, b"hi", bytearray(b"hi")]:
    print(f"{value}  →  {type(value).__name__}")
````

### 3.5 Type Conversion and Type Checking
Conversion is done with the type name as a function: `int("42")`, `str(42)`, `list("abc")`, and so on. Checking is best done with `isinstance()` rather than comparing `type(x) == int`, because `isinstance` respects subclasses.

````python
print(int("42"), float("3.14"), str(42), bool(0), bool(1), list("abc"))
print(isinstance(42, int))        # True
print(isinstance(True, int))      # True — bool inherits from int, useful when you least expect it
````

### 3.6 Mutable versus Immutable Objects
Half of the surprises in Python come down to this distinction.

- **Immutable:** `int`, `float`, `str`, `tuple`, `bytes`, `frozenset`. Once created, the object never changes. You get a new object every time you "modify" it.
- **Mutable:** `list`, `dict`, `set`, `bytearray`. The object can change in place; every name pointing at it sees the change.

````python
s = "hello"
try:
    s[0] = "H"
except TypeError as e:
    print("strings are immutable:", e)

lst = [1, 2, 3]
lst[0] = 99                     # lists are mutable — this works
print("list mutated in place:", lst)
````

### 3.7 Truthy and Falsy Values
Every value in Python has a truthiness. In an `if`, `while`, or `bool()` context, these are **falsy**: `0`, `0.0`, `""`, `[]`, `{}`, `()`, `set()`, `None`, `False`. Everything else — including `"False"` (a non-empty string) and `[0]` (a list that happens to contain a falsy element) — is truthy.

````python
for v in [0, 0.0, "", [], {}, None, False, "text", 1, [0], "False"]:
    print(f"{v} → bool = {bool(v)}")
````

## 4. Operators

### 4.1 Arithmetic Operators
| Op | Meaning | Note |
|----|---------|------|
| `+` `-` `*` | usual | |
| `/` | true division | always returns `float` |
| `//` | floor division | integer if both operands are int |
| `%` | remainder (modulo) | |
| `**` | exponent | |

````python
print(7 / 2)     # 3.5   — float
print(7 // 2)    # 3     — floor
print(7 % 2)     # 1     — remainder
print(2 ** 10)   # 1024
````

### 4.2 Comparison Operators
`==`, `!=`, `<`, `>`, `<=`, `>=` all return a `bool`. Python lets you chain comparisons the way you would write them in math — much more readable than the C-style two-condition form.

````python
a = 5
print(1 < a < 10)      # True — reads exactly like the math
print(a == 5.0)        # True — Python compares numeric values across int and float
````

### 4.3 Logical Operators: `and`, `or`, and `not`
Two things to remember. First, they short-circuit — `and` stops at the first falsy operand and returns it; `or` stops at the first truthy operand and returns it. Second, they don't necessarily return a bool — they return one of the operands.

````python
print(True and False)
print(True or False)
print(not True)

# the classic default-value idiom
print(True or "anonymous")
print(False or "anonymous")
print(True and "anonymous")
print(False and "anonymous")
````

### 4.4 Augmented Assignment Operators
`+=`, `-=`, `*=`, `/=`, `//=`, `%=`, and `**=` combine an operation with assignment. For immutable numbers, `counter += 3` behaves like `counter = counter + 3`. With mutable objects, augmented assignment may modify the existing object in place.

````python
counter = 10
counter += 3
print(counter)   # 13
counter *= 2
print(counter)   # 26
counter //= 4
print(counter)   # 6
counter **= 2
print(counter)   # 36
counter %= 100
print(counter)   # 36
````

### 4.5 Identity Operators: `is` and `is not`
This is the operator most likely to surprise you. `==` asks *are these two objects equal?*; `is` asks *are these two names pointing at the same object?* They are not interchangeable.

````python
a = [1, 2]
b = [1, 2]
c = a
print(a == b)      # True   — same value
print(a is b)      # False  — different objects
print(a is c)      # True   — c is another name for the same object
print(a is not b)  # True
````

### 4.6 Membership Operators: `in` and `not in`
Works on strings, lists, tuples, sets, and dicts. On a dict, `in` tests **keys**, not values.

````python
print("a" in "cat")                       # True - 'a' is a substring
print(3 in [1, 2, 3])                     # True - 3 is an element in the list
print("key" in {"key": "val"})            # True — dict membership tests keys
print("val" in {"key": "val"})            # False — dict membership tests keys
print("val" in {"key": "val"}.values())   # True — dict membership values check
````

## 5. Strings

DevOps is 80% text — logs, configs, YAML, JSON, output of shell commands. Getting comfortable with strings pays off.

### 5.1 Quote Styles
Three ways to write a string literal. Pick whichever lets you avoid escaping. Triple quotes let you span multiple lines without any escapes.

````python
a = 'single'
b = "double"
c = """triple
spans
lines"""
print(a, b)
print(c)
print("O'Brien")         # double avoids escaping the apostrophe
print('O\'Brien')        # Escaping is required when a single-quote appears inside a single-quoted string:
````

### 5.2 Indexing and Slicing
Zero-based. Negative indices count from the end. Slice syntax is `s[start:stop:step]` and `stop` is **exclusive**. This bites everyone at least once.

````python
s = "hello"
print(s[0], s[-1])       # 'h' 'o'
print(s[1:4])            # 'ell'
print(s[:2], s[2:])      # 'he' 'llo'
print(s[::-1])           # 'olleh' — reversed
````

### 5.3 Common String Methods
Every string method returns a **new** string — strings are immutable, remember. That is why we can safely chain them.

````python
s = "  Bytown Data Center  "
print(s.strip())
print(s.strip().upper())
print(s.strip().lower())
print(s.strip().title())
print(s.strip().replace("Data", "DATA"))
print("a,b,c".split(","))
print("-".join(["a", "b", "c"]))
print(",".join(["a", "b", "c"]))
print("config.yaml".endswith(".yaml"))
print("config.yaml".startswith("config"))
print("error error warn".count("error"))
print("hello world".find("world"))
````

### 5.4 String Formatting — Three Approaches
You will meet all three in the wild. f-strings are the modern default and by far the most readable. `.format()` was the standard between roughly 2010 and 2016 and still shows up in mature codebases. `%`-style is the oldest and lives on inside log lines and any C-derived tooling.

````python
name, ports = "sw01", 24
print("%s has %d ports" % (name, ports))          # %-style
print("{} has {} ports".format(name, ports))       # .format()
print(f"{name} has {ports} ports")                 # f-string
````

### 5.5 Escape Characters
| Escape | Meaning |
|--------|---------|
| `\n` | newline |
| `\t` | tab |
| `\\` | literal backslash |
| `\"` | literal double quote |
| `\'` | literal single quote |

````python
print("line1\nline2")
print("col1\tcol2")
print("path with \\ backslash")
print("path with \"double\" quotes")
print('path with \'single\' quotes')
````

### 5.6 Raw Strings
A raw string, prefixed with `r`, treats backslashes as literal characters. Two places you will reach for this constantly: **Windows paths** and **regular expressions**.

````python
print("C:\\Users\\Amir\\logs\\today.txt")   # normal string — every backslash escaped
print(r"C:\Users\Amir\logs\today.txt")       # raw string — much easier to read
````

### 5.7 Unicode: `encode()` and `decode()`
Python 3 strings contain Unicode text. Files, network protocols, and subprocesses exchange bytes, commonly encoded as UTF-8. Use `encode()` to convert text to bytes and `decode()` to convert bytes back to text.

````python
emoji = "\u2713"           # ✓
b = emoji.encode("utf-8")  # b'\xe2\x9c\x93'
back = b.decode("utf-8")   # ✓
print(emoji)
print(b)
print(back)
````

## 6. Lists

The workhorse. Ordered, mutable, allow duplicates, index by integer.

````python
x = [10, 20, 30, 40]
print(x[0], x[-1])       # 10 40
print(x[1:3])            # [20, 30]
x[0] = 99                 # lists are mutable
print(x)
````

### 6.1 Methods That Mutate in Place
These change the list itself. Note the return value of `pop` — it hands you the element it removed.

````python
x = ["GigE0/1", "GigE0/2"]
x.append("GigE0/3")
x.extend(["Mgmt0", "Console"])
x.insert(0, "Loopback0")
print(x)
x.remove("Console")
last = x.pop()
print("popped:", last, "remaining:", x)
````

### 6.2 Methods That Inspect a List

````python
x = ["a", "b", "a", "c"]
print(x.index("b"))      # position of first 'b'
print(x.count("a"))      # how many 'a's
````

### 6.3 Sorting, Reversing, and Copying
`sorted(x)` returns a new sorted list, while `x.sort()` sorts the existing list in place. `reversed(x)` returns a reverse iterator, while `x.reverse()` reverses the existing list in place. Use `x.copy()` when you need an independent shallow copy before mutation.

````python
x = [3, 1, 2]
print(sorted(x), x)
copy = x.copy()
copy.reverse()
print(copy, x)
````

### 6.4 Nested Lists
Lists can hold anything, including other lists. A 2D grid is just a list of rows.

````python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(matrix[1])
print(matrix[1][2])
````

## 7. Tuples and Sets

### 7.1 Tuples — Ordered, Immutable, and Allow Duplicates
Reach for a tuple when the group has a fixed size and shape (a record like `hostname, ip, ports`) or when you need something you can use as a dictionary key — lists cannot be keys because they are mutable.

````python
dev = ("SW-CORE-02", "10.10.4.12", 48)
print(dev[0], dev[-1])

# Tuple items are immutable, unless you update the entire tuple
try:
    dev[0] = "X"
except TypeError as e:
    print("tuples are immutable:", e)

dev = ("X", dev[1], dev[2])
print(dev)
````

### 7.2 Tuple Unpacking
Anywhere Python expects a tuple, you can unpack it into named variables in one step. This is why loops over dictionary items read so cleanly.

````python
hostname, ip, ports = dev
print(hostname, ip, ports)
````

### 7.3 Named Tuples — Records with Attribute Access
For records used in more than one place, a named tuple gives you all the immutability of a regular tuple plus attribute access by name. Much easier to read than `d[2]`.

````python
from collections import namedtuple
Device = namedtuple("Device", ["hostname", "ip", "ports"])
d = Device("SW-CORE-02", "10.10.4.12", 48)
print(d)
print(d.hostname, d.ports)
````

### 7.4 Sets — Unordered Collections of Unique Elements
Sets eliminate duplicates automatically, provide average constant-time membership tests, and support mathematical set operations. They are useful for comparing packages installed on two servers or determining which required ports are open.

````python
packages = {"nginx", "openssh", "python3", "vim", "nginx"}   # duplicate collapses
print(packages)
packages.add("curl")
packages.remove("vim")
print(packages)
````

### 7.5 Set Operations
Read these left to right. `|` union, `&` intersection, `-` difference, `^` symmetric difference. `<=` and `>=` for subset/superset.

````python
a = {"nginx", "openssh", "python3", "vim"}
b = {"nginx", "openssh", "curl", "git"}
print("union       :", a | b)
print("intersect   :", a & b)
print("a - b       :", a - b)
print("b - a       :", b - a)
print("symmetric   :", a ^ b)
print("{nginx} ⊆ a :", {"nginx"} <= a) # {nginx} is a subset of a
print("a ⊇ {nginx}:", a >= {"nginx"})  # a is a superset of {nginx}
````

## 8. Dictionaries

A dictionary maps unique, **hashable** keys to values. Common key types include strings, numbers, and tuples containing only hashable elements. Dictionaries preserve insertion order in Python 3.7 and later.

Almost every JSON payload, YAML config, and API response you touch is a dictionary. Get comfortable.

````python
d = {"host": "sw01", "ports": 24}
print(d["host"])
d["status"] = "on"       # add
d["ports"] = 48           # overwrite
del d["status"]           # remove
print(d)
````

### 8.1 Safe Access with `get()` and `setdefault()`
Accessing a missing key with square brackets raises `KeyError`. `get()` returns a default instead. `setdefault()` returns the value if the key exists, or inserts your default and returns *that*. Very handy when you want to append to a list that might not exist yet.

````python
d = {"host": "sw01"}
print(d.get("host"))                # 'sw01'
print(d.get("vlan", "unknown"))     # 'unknown' — no KeyError
tags = d.setdefault("amir", [])     # creates the list if missing, returns it
tags.append("core")
print(d)
tags.append("data")
print(d)
````

### 8.2 `pop()`, `popitem()`, and `update()`
- `pop(key)` removes the key and returns its value.
- `popitem()` removes and returns the **most recently inserted** item as a `(key, value)` tuple.
- `update(other)` merges another dict (or an iterable of pairs) into this one.

````python
d = {"host": "sw01", "ports": 24, "vlan": 10}
print(d)
print(d.pop("vlan"))            # 10
print(d)
print(d.popitem())               # ('ports', 24)
print(d)
d.update({"status": "on", "ports": 48})
print(d)
````

### 8.3 Iteration with `keys()`, `values()`, and `items()`
Most of the time you want `items()`. It gives you both the key and the value in one clean loop.

````python
inventory = {
    "sw01": {"ip": "10.0.0.1", "ports": 48},
    "sw02": {"ip": "10.0.0.2", "ports": 24},
}

for name, info in inventory.items():
    print(f"{name} → {info['ip']}")

print(list(inventory.keys()))
print(list(inventory.values()))
````

## 9. Control Flow

### 9.1 `if`, `elif`, and `else`
The first matching branch wins — the rest are skipped. This means **the order of your conditions matters** when they overlap.

````python
for x in [3, 12, 250]:
    if x < 5:
        print(x, "small")
    elif x < 100:
        print(x, "medium")
    else:
        print(x, "large")
````

### 9.2 Conditional Expression
`value_if_true if condition else value_if_false`. Use it for simple binary choices; anything more complex is clearer as a full `if / else`.

````python
n = 42
label = "even" if n % 2 == 0 else "odd"
print(label)
````

### 9.3 `for` and `range()`
A `for` loop in Python iterates over a sequence, not a counter. If you want a counter, ask for it explicitly with `range()`.

````python
for i in range(5):         # 0..4
    print(i, end=" ")
print()
for i in range(2, 8):      # 2..7
    print(i, end=" ")
print()
for i in range(0, 10, 2):  # 0, 2, 4, 6, 8
    print(i, end=" ")
print()
````

### 9.4 `enumerate()` — Index and Value Together
If you find yourself writing `for i in range(len(x))`, stop and use `enumerate` instead. It is one of the small habits that separates readable Python from code translated straight out of Java.

````python
for i, v in enumerate(["nginx", "openssh", "curl"]):
    print(i, v)
````

### 9.5 `while` Loops
Use `while` when you do not know in advance how many iterations you need. Otherwise, prefer `for`.

````python
n = 32
iterations = 0
while n >= 5:
    n //= 2
    iterations += 1
    print(f"iter {iterations}: n = {n}")
````

### 9.6 `break`, `continue`, and `pass`
- `break` exits the current loop entirely.
- `continue` skips to the next iteration.
- `pass` does nothing. It's a placeholder for spots where Python's syntax requires a statement but you have nothing to put there yet.

````python
for i in range(10):
    if i > 6:
        break             # stop looping altogether
    if i % 2 == 1:
        continue          # skip odd numbers
    if i == 4:
        pass              # do nothing special; keep going
    print(i)
````

> **Common pitfall.** In a chain of `elif`s the first true branch wins, so order matters. If your rules overlap, put the most specific one first.

## 10. Functions

The unit of reuse. If you write the same three lines twice, put them in a function.

### 10.1 Function Definition and Docstring
The `def` keyword, a name, parentheses, arguments, colon, body. If the first thing in the body is a triple-quoted string, Python treats it as the function's docstring — a real object attribute accessible at runtime.

````python
def greet(name):
    """Return a greeting string for the given name."""
    return f"Hello, {name}!"

print(greet("Amir"))
print(greet.__doc__)
````

### 10.2 Positional, Keyword, and Default Arguments
Callers can pass arguments by position or by name. Parameters can have defaults, which makes them optional at the call site.

````python
def summarize(host, ports, online=True):
    return f"{host} · {ports} ports · online={online}"

print(summarize("sw01", 24))                       # positional with default value
print(summarize("sw01", 24, False))                # positional with overwritten default value
print(summarize(host="sw01", ports=24))            # keyword
print(summarize("sw01", 24, online=False))         # mix
````

### 10.3 Variable-Length Arguments: `*args` and `**kwargs`
When you don't know how many positional or keyword arguments a caller will pass, `*args` and `**kwargs` catch the rest. The names are convention — what matters is the `*` and `**`.

````python
def total(*nums):           # *args collects positional args into a tuple
    return sum(nums)

def config(**opts):         # **kwargs collects keyword args into a dict
    return opts

print(total(1, 2, 3, 4))
print(config(host="sw01", ports=24))
````

### 10.4 Multiple Return Values
Python doesn't actually return multiple values — it returns one tuple. But the syntax makes it feel natural.

````python
def minmax(nums):
    return min(nums), max(nums)     # returns a tuple

lo, hi = minmax([3, 1, 4, 1, 5, 9, 2])
print(lo, hi)
````

### 10.5 Type Hints
Optional annotations of what each argument and the return value are expected to be. Python does not enforce them at runtime — they are documentation for humans and static checkers like mypy or pyright. Very much worth adopting.

````python
def area(radius: float) -> float:
    """Return the area of a circle."""
    return 3.14 * radius ** 2

print(area(5))
````

### 10.6 Scope — `global` and `nonlocal`
Names inside a function are local by default. To *write* to a name in an enclosing scope, you need to say so. `global` targets the module-level name; `nonlocal` targets the nearest enclosing function's name.

````python
COUNT = 0

def bump():
    global COUNT       # write to the module-level name
    COUNT += 1

def make_counter():
    n = 0
    def inner():
        nonlocal n     # write to the enclosing function's name (not global)
        n += 1
        return n
    return inner

bump()
bump()
print("COUNT =", COUNT)

next_id = make_counter()
print(next_id(), next_id(), next_id())
````

### 10.7 Recursion
A function calling itself. Needs a **base case** or it will recurse forever and eventually crash the interpreter with `RecursionError`. Python's default recursion limit is around 1000 — deep enough for most sensible cases, but you would not use recursion for iterating a million-element list.

````python
def countdown(n):
    print(n)
    if n > 1:
        countdown(n - 1)
    else:
        print("go")

countdown(3)
````
